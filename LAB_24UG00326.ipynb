```python
# Copy original
df_clean = df.copy()

# Time consistency
df_clean = df_clean[df_clean['tpep_dropoff_datetime'] > df_clean['tpep_pickup_datetime']]

# Distance
df_clean = df_clean[df_clean['trip_distance'] > 0]

# Passenger
df_clean = df_clean[(df_clean['passenger_count'] > 0) & (df_clean['passenger_count'] <= 6)]

# Fare
df_clean = df_clean[df_clean['fare_amount'] >= 0]

# Total
df_clean = df_clean[df_clean['total_amount'] > 0]

# Duration
df_clean['trip_duration'] = (
    df_clean['tpep_dropoff_datetime'] - df_clean['tpep_pickup_datetime']
).dt.total_seconds() / 60

df_clean = df_clean[(df_clean['trip_duration'] > 1) & (df_clean['trip_duration'] < 180)]

# Speed
df_clean['speed'] = df_clean['trip_distance'] / (df_clean['trip_duration'] / 60)

df_clean = df_clean[(df_clean['speed'] > 2) & (df_clean['speed'] < 100)]
```

When the definition of a “real trip” becomes stricter, the system appears more stable. When relaxed, variability increases, revealing hidden inefficiencies.

THIS connects to Act 1:

instability zone = Weeks 1–3
but depends on definition of “real”

```python
df_analysis = df_clean.copy()

# Extract week
df_analysis['week'] = df_analysis['tpep_pickup_datetime'].dt.isocalendar().week

# Tip ratio (behavioral metric)
df_analysis['tip_ratio'] = df_analysis['tip_amount'] / df_analysis['fare_amount']

# Remove extreme ratios
df_analysis = df_analysis[(df_analysis['tip_ratio'] >= 0) & (df_analysis['tip_ratio'] <= 1)]
```

```python
stress = df_analysis[(df_analysis['week'] >= 1) & (df_analysis['week'] <= 3)]
normal = df_analysis[(df_analysis['week'] > 3)]
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(stress['tip_ratio'], normal['tip_ratio'], equal_var=False)

print("T-stat:", t_stat)
print("P-value:", p_value)
```

The p-value obtained (~2.17 × 10⁻³¹) is significantly lower than the standard threshold of 0.05, leading to rejection of the null hypothesis. This indicates a statistically significant difference in tipping behavior between the stress period (Weeks 1–3) and the normal period.                                Your T-stat is negative
That means:
Mean(stress tip_ratio) < Mean(normal tip_ratio)

```python
print("Stress mean:", stress['tip_ratio'].mean())
print("Normal mean:", normal['tip_ratio'].mean())
```

The negative t-statistic indicates that tipping ratios during the stress period are lower than during normal periods, suggesting a decline in passenger generosity or willingness to tip under unstable system conditions.

```python
from scipy.stats import ttest_ind

t_stat_dist, p_val_dist = ttest_ind(stress['trip_distance'], normal['trip_distance'], equal_var=False)

print("Distance P-value:", p_val_dist)
```

The statistical test on trip distance also shows a highly significant difference (p ≈ 0), indicating that trip composition varies between stress and normal periods. This suggests that changes in tipping behavior may partially result from differences in trip types rather than purely behavioral factors.

While trip distance influences tipping behavior, the use of a normalized metric (tip_ratio) reduces this dependency. The continued presence of statistically significant differences suggests that behavioral factors play a meaningful role alongside mechanical variations.

Therefore, the system exhibits both structural and behavioral changes during the stress period.

ACT 5: Chosen Parameter: Fare (via fare_amount)

Justification:
Fare directly influences passenger behavior (demand, tipping) and driver incentives. A small change in fare can propagate through the system, affecting both economic and behavioral dynamics. Therefore, it is an effective parameter to test system sensitivity.

```python
R = 2
change_factor = 1 + (R / 100)   # 1.02 → 2% increase
```

```python
df_sim = df_analysis.copy()

# Apply fare increase
df_sim['fare_new'] = df_sim['fare_amount'] * change_factor

# Recalculate tip ratio (behavior affected)
df_sim['tip_ratio_new'] = df_sim['tip_amount'] / df_sim['fare_new']
```

```python
print("Original Fare Mean:", df_analysis['fare_amount'].mean())
print("New Fare Mean:", df_sim['fare_new'].mean())

print("Original Tip Ratio Mean:", df_analysis['tip_ratio'].mean())
print("New Tip Ratio Mean:", df_sim['tip_ratio_new'].mean())
```

This indicates that passengers do not proportionally adjust tipping when fares increase, suggesting a bounded behavioral response where tipping does not scale linearly with price changes.
